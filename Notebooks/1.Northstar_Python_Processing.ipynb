#**Python Data Processing**

##**1. Import & Setup**

```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display Settings
sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (10, 5)

print("Libraries loaded")
```

## **2. Load all datasets**

```python
#The datasets were loaded directly from GitHub repository URLs into Pandas DataFrames
customers = pd.read_csv('https://raw.githubusercontent.com/Habishalini/NorthStar-Database_Analytics-Coursework/refs/heads/main/Datasets/customers.csv')
drivers = pd.read_csv('https://raw.githubusercontent.com/Habishalini/NorthStar-Database_Analytics-Coursework/refs/heads/main/Datasets/drivers.csv')
vehicles = pd.read_csv('https://raw.githubusercontent.com/Habishalini/NorthStar-Database_Analytics-Coursework/refs/heads/main/Datasets/vehicles.csv')
orders = pd.read_csv('https://raw.githubusercontent.com/Habishalini/NorthStar-Database_Analytics-Coursework/refs/heads/main/Datasets/orders.csv')
deliveries = pd.read_csv('https://raw.githubusercontent.com/Habishalini/NorthStar-Database_Analytics-Coursework/refs/heads/main/Datasets/deliveries.csv')
complaints = pd.read_csv('https://raw.githubusercontent.com/Habishalini/NorthStar-Database_Analytics-Coursework/refs/heads/main/Datasets/complaints.csv')
incidents = pd.read_csv('https://raw.githubusercontent.com/Habishalini/NorthStar-Database_Analytics-Coursework/refs/heads/main/Datasets/incidents.csv')
hubs = pd.read_csv('https://raw.githubusercontent.com/Habishalini/NorthStar-Database_Analytics-Coursework/refs/heads/main/Datasets/hubs.csv')
app_events = pd.read_csv('https://raw.githubusercontent.com/Habishalini/NorthStar-Database_Analytics-Coursework/refs/heads/main/Datasets/app_events.csv')

print("All datasets loaded")
```

## **3. Data Overview**

```python
# Display shape and first 5 rows for all datasets
datasets = {
    "customers": customers,
    "drivers": drivers,
    "vehicles": vehicles,
    "orders": orders,
    "deliveries": deliveries,
    "complaints": complaints,
    "incidents": incidents,
    "hubs": hubs,
    "app_events": app_events
}

for name, df in datasets.items():
    print(f"\n{name.upper()} Dataset")
    print("Shape:", df.shape)
    display(df.head())
```

## **4. Data Quality Check (Before Cleaning)**

```python
quality_before = []

for name, df in datasets.items():
    quality_before.append({
        "Dataset": name,
        "Missing Values": df.isnull().sum().sum(),
        "Duplicate Rows": df.duplicated().sum()
    })

quality_before_df = pd.DataFrame(quality_before)
quality_before_df
```

## **5. Check missing values**

```python
# Check missing values in all datasets
for name, df in datasets.items():
    print(f"\n{name.upper()}")

    missing = df.isnull().sum()

    if missing.sum() == 0:
        print("No missing values")
    else:
        print(missing[missing > 0])
```

## **6. Fill missing values**

```python
# Fill numeric columns with mean, text columns with mode
def fill_missing(df):
    df = df.copy()
    numeric_cols = df.select_dtypes(include="number").columns
    object_cols  = df.select_dtypes(exclude="number").columns
    for col in numeric_cols:
        if df[col].isnull().any():
            df[col] = df[col].fillna(df[col].mean())
    for col in object_cols:
        if df[col].isnull().any():
            df[col] = df[col].fillna(df[col].mode().iloc[0])
    return df

# Apply to each variable directly
customers   = fill_missing(customers)
drivers     = fill_missing(drivers)
vehicles    = fill_missing(vehicles)
orders      = fill_missing(orders)
deliveries  = fill_missing(deliveries)
complaints  = fill_missing(complaints)
incidents   = fill_missing(incidents)
hubs        = fill_missing(hubs)
app_events  = fill_missing(app_events)

# Check results
total_nulls = sum(df.isnull().sum().sum() for df in
    [customers, drivers, vehicles, orders, deliveries, complaints, incidents, hubs, app_events])
print(f"Total nulls remaining: {total_nulls}")
```

## **7. Convert date columns**

```python
# Convert date columns to datetime format

customers["signup_date"] = pd.to_datetime(customers["signup_date"])
vehicles["commission_date"] = pd.to_datetime(vehicles["commission_date"])
orders["order_created_at"] = pd.to_datetime(orders["order_created_at"])
deliveries["dispatch_time"] = pd.to_datetime(deliveries["dispatch_time"])
deliveries["delivery_completed_at"] = pd.to_datetime(deliveries["delivery_completed_at"])
complaints["created_at"] = pd.to_datetime(complaints["created_at"])
incidents["reported_at"] = pd.to_datetime(incidents["reported_at"])
app_events["event_timestamp"] = pd.to_datetime(app_events["event_timestamp"])

print("Date columns converted")
```

## **8. Check duplicate values**

```python
for name, df in datasets.items():

    dup_count = df.duplicated().sum()

    print(f"{name}: {dup_count} duplicate rows")
```

## **9. Check inconsistent categorical values**

```python
for name, df in datasets.items():

    print(f"\nDATASET: {name}")
    print("-" * 50)

    categorical_cols = df.select_dtypes(include='object').columns

    for col in categorical_cols:

        unique_values = df[col].dropna().unique()

        print(f"\nColumn: {col}")
        print(f"Unique values count: {len(unique_values)}")

        # Show only first 10 values
        print("Sample values:", unique_values[:10])

        print("-" * 30)
```

## **10. Fix inconsistent zone names**

```python
# Fix inconsistent zone names

zone_fix = {"Ctr": "Central"}

def clean_zone(column):

    column = column.str.strip()
    column = column.str.title()
    column = column.replace(zone_fix)

    return column

# Apply cleaning to all zone columns
customers["home_zone"] = clean_zone(customers["home_zone"])
drivers["base_zone"] = clean_zone(drivers["base_zone"])
vehicles["assigned_zone"] = clean_zone(vehicles["assigned_zone"])
orders["pickup_zone"] = clean_zone(orders["pickup_zone"])
orders["dropoff_zone"] = clean_zone(orders["dropoff_zone"])
hubs["zone"] = clean_zone(hubs["zone"])
app_events["zone_context"] = clean_zone(app_events["zone_context"])


# Check results
print("Customer home_zone values AFTER cleaning:")
print(customers["home_zone"].unique())
```

## **11. Create final cleaned dataset dictionary**

```python
datasets = {
    "customers": customers,
    "drivers": drivers,
    "vehicles": vehicles,
    "orders": orders,
    "deliveries": deliveries,
    "complaints": complaints,
    "incidents": incidents,
    "hubs": hubs,
    "app_events": app_events
}

print("Data cleaning complete. Ready for analysis.")
```

## **12. Data Quality Check (After Cleaning)**

```python
quality_after = []
for name, df in datasets.items():
    quality_after.append({
        "Dataset": name,
        "Missing Values": df.isnull().sum().sum(),
        "Duplicate Rows": df.duplicated().sum()
    })
quality_after_df = pd.DataFrame(quality_after)
print(quality_after_df)
```

## **13. Combining Deliveries and Order Data**
Merge deliveries with orders to get pickup zone.



```python
delivery_order = pd.merge(deliveries, orders, on="order_id", how="inner")
print("Shape:", delivery_order.shape)
delivery_order.head()
```

## **14. Feature engineering**
Create new columns from existing data

```python
# Profit per delivery
delivery_order["profit"] = delivery_order["order_value"] - delivery_order["fuel_or_charge_cost"]

# Check is the delivery failed? (1 = yes, 0 = no)
delivery_order["is_failed"] = 0
delivery_order.loc[delivery_order["delivery_status"] == "Failed", "is_failed"] = 1

# Check is the delivery late? (failed OR delayed)
delivery_order["is_late"] = 0
delivery_order.loc[delivery_order["delivery_status"] == "Failed", "is_late"] = 1
delivery_order.loc[delivery_order["delivery_status"] == "Delayed", "is_late"] = 1

# Check driver override the route?
delivery_order["has_override"] = 0
delivery_order.loc[delivery_order["manual_route_override_count"] > 0, "has_override"] = 1

# Time features
delivery_order["actual_duration_hrs"] = (delivery_order["delivery_completed_at"] - delivery_order["dispatch_time"]).dt.total_seconds() / 3600
delivery_order["dispatch_hour"] = delivery_order["dispatch_time"].dt.hour
delivery_order["dispatch_day"] = delivery_order["dispatch_time"].dt.day_name()

# Check order have a complaint?
complaint_orders = complaints["order_id"].unique()
delivery_order["has_complaint"] = 0
delivery_order.loc[delivery_order["order_id"].isin(complaint_orders), "has_complaint"] = 1

# Show first few rows
print(delivery_order[["delivery_status", "is_failed", "is_late", "has_override", "profit"]].head())
```

## **15. Chart 1 - Delivery status by pickup zone**

```python
status_by_zone = pd.crosstab(delivery_order["pickup_zone"], delivery_order["delivery_status"])

status_by_zone.plot(kind="bar", figsize=(10, 5))
plt.title("Delivery Status by Pickup Zone")
plt.xlabel("Pickup Zone")
plt.ylabel("Number of Deliveries")
plt.xticks(rotation=0)
plt.show()

print("Worst zone:", status_by_zone["Failed"].idxmax(),
      "with", status_by_zone["Failed"].max(), "failures")
```

## **16. Chart 2 - Top 10 drivers by route overrides**
Driver info attached to delivery rows to group by driver.

```python
delivery_driver = pd.merge(deliveries, drivers, on="driver_id", how="inner")

top10 = (delivery_driver.groupby("driver_id")["manual_route_override_count"]
                .mean()
                .sort_values(ascending=False)
                .head(10))

top10.plot(kind="bar", color="red", figsize=(8, 3))
plt.title("Top 10 Drivers by Route Overrides")
plt.xlabel("Driver ID")
plt.ylabel("Avg Overrides per Delivery")
plt.xticks(rotation=45)
plt.show()

print("Worst driver:", top10.idxmax())
```

## **17. Chart 3 - Maintenance status vs incidents**
 Incident links to a delivery, the delivery links to a vehicle, and the vehicle has the maintenance status.

```python
m = pd.merge(incidents, deliveries[["delivery_id", "vehicle_id"]], on="delivery_id")
m = pd.merge(m, vehicles[["vehicle_id", "maintenance_status"]], on="vehicle_id")

inc_by_maint = m["maintenance_status"].value_counts()

inc_by_maint.plot(kind="bar", color="steelblue", figsize=(8, 4))
plt.title("Incidents by Maintenance Status")
plt.xlabel("Maintenance Status")
plt.ylabel("Number of Incidents")
plt.xticks(rotation=0)
plt.show()

print("Most incidents:", inc_by_maint.idxmax())
```

## **18. Chart 4 - Most common complaint type**

```python
c = complaints["complaint_type"].value_counts()

c.plot(kind="bar", color="orange", figsize=(10, 4))
plt.title("Complaints by Type")
plt.xlabel("Complaint Type")
plt.ylabel("Number of Complaints")
plt.xticks(rotation=30)
plt.show()

print("Most common:", c.idxmax(), "with", c.max(), "cases")
```

## **19. Chart 5 - Average profit per zone**

```python
profit_by_zone = (delivery_order.groupby("pickup_zone")["profit"]
                          .mean().round(2)
                          .sort_values(ascending=False))

profit_by_zone.plot(kind="bar", color="green", figsize=(8, 4))
plt.title("Average Profit per Delivery by Zone")
plt.xlabel("Pickup Zone")
plt.ylabel("Avg Profit (£)")
plt.xticks(rotation=0)
plt.show()

print(f"\nHighest profit zone: {profit_by_zone.idxmax()} = £{profit_by_zone.max()}")
print(f"Lowest profit zone: {profit_by_zone.idxmin()} = £{profit_by_zone.min()}")
```

## **20. Heatmap 1 - complaint severity by zone**

```python
cz = pd.merge(complaints, orders[["order_id", "pickup_zone"]], on="order_id")
sev = pd.crosstab(cz["pickup_zone"], cz["severity"])

plt.figure(figsize=(8, 5))
sns.heatmap(sev, annot=True, fmt="d", cmap="Reds")
plt.title("Complaint Severity by Zone")
plt.xlabel("Severity")
plt.ylabel("Pickup Zone")
plt.show()
```

## **21. Heatmap 2 - Service Efficiency by Zone**

```python
zone_service = (delivery_order.groupby(["pickup_zone", "service_type"])["is_late"]
                       .mean()
                       .unstack()
                       .round(3) * 100).round(1).fillna(0)

plt.figure(figsize=(8, 4))
sns.heatmap(zone_service, annot=True, fmt=".1f", cmap="RdYlGn_r",
            linewidths=0.5, cbar_kws={"label": "Failure/Delay Rate (%)"})
plt.title("Failure/Delay Rate: Zone × Service Type")
plt.tight_layout()
plt.show()
```

## **22. Failure rate by hour of dispatch**

```python
hourly = delivery_order.groupby("dispatch_hour").agg(
    total=("delivery_id", "count"),
    late_count=("is_late", "sum")
)
hourly["fail_pct"] = (hourly["late_count"] / hourly["total"] * 100).round(1)

# Calculate the average failure rate
avg_rate = hourly["fail_pct"].mean()
print("Average failure rate across all hours:", round(avg_rate, 1), "%")

plt.figure(figsize=(8, 5))
plt.bar(hourly.index, hourly["fail_pct"], color="tomato", edgecolor="white")
plt.axhline(avg_rate, color="navy", linestyle="--", label="Daily average")
plt.title("Failure / Delay Rate by Hour of Dispatch")
plt.xlabel("Hour (0 = midnight, 12 = noon)")
plt.ylabel("Failure / Delay Rate (%)")
plt.xticks(range(24))
plt.legend()
plt.tight_layout()
plt.show()

print("\n3 Worst Hours (highest failure rate):")
worst_hours = hourly.nlargest(3, "fail_pct")[["fail_pct"]]
for hour, row in worst_hours.iterrows():
    print(f"  Hour {hour}: {row['fail_pct']}% failure rate")

print("\n3 Best Hours (lowest failure rate):")
best_hours = hourly.nsmallest(3, "fail_pct")[["fail_pct"]]
for hour, row in best_hours.iterrows():
    print(f"  Hour {hour}: {row['fail_pct']}% failure rate")
```

## **23. App performance analysis**

```python
# Calculate app performance per event type
avg_latency = app_events.groupby("event_type")["api_latency_ms"].mean().round(2)
success_rate = app_events.groupby("event_type")["success_flag"].mean().round(3)
event_count = app_events.groupby("event_type")["event_id"].count()

# Combine into one table
app_perf = pd.DataFrame({
    "avg_latency": avg_latency,
    "success_rate": success_rate,
    "count": event_count
}).sort_values("avg_latency", ascending=False)

print(app_perf)

# Two charts side by side
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].barh(app_perf.index, app_perf["avg_latency"], color="steelblue")
axes[0].axvline(300, color="red", linestyle="--", label="300ms target")
axes[0].set_title("Average API Response Time (ms)")
axes[0].set_xlabel("Milliseconds")
axes[0].legend()

axes[1].barh(app_perf.index, app_perf["success_rate"] * 100, color="green")
axes[1].set_title("App Success Rate (%)")
axes[1].set_xlabel("%")
axes[1].set_xlim(85, 100)

plt.tight_layout()
plt.show()
```

## **24. Zone summary table**
Combines failure rate, rating and profit per zone in one table.

```python
summary = pd.DataFrame({
    "failed": status_by_zone["Failed"],
    "failure_rate_%": (status_by_zone["Failed"] / status_by_zone.sum(axis=1) * 100).round(2),
    "avg_rating": delivery_order.groupby("pickup_zone")["customer_rating_post_delivery"].mean().round(2),
    "avg_profit": profit_by_zone
}).sort_values("failure_rate_%", ascending=False)

display(summary)
```
